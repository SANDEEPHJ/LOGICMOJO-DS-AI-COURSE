```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
## reading csv file : data ##
df=pd.read_csv("Banking_churn_prediction.csv")
```

```python
## showing first row ##
print(df.head)
```

```python
## showing last row ##
print (df.tail)
```

```python
## data shape no. of rows & columns##
print("rows and columns:",df.shape)
```

```python
## column names : header and data type##
print("\n column name")
print(df.columns)

print ("\n data types:")
print (df.dtypes)
```

```python
## Data summary ##
print ("\n summary")
print(df.describe())
```

```python
## identify missing values##
print("\n missing values")
print(df.isnull().sum())
```

```python
print(df['churn'].value_counts())
plt.figure(figsize=(4,5))

sns.countplot(x='churn', data=df)
plt.title("Churn Distribution")
plt.show()
```

```python
plt.figure(figsize=(8,8))
sns.histplot(df['age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.histplot(df['current_balance'], bins=30)
plt.title("Current Balance Distribution")
plt.show()
```

```python
sns.countplot(x='gender', hue='churn', data=df)
plt.title("Gender vs Churn")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(x='occupation', hue='churn', data=df)
plt.xticks(rotation=45)
plt.title("Occupation vs Churn")
plt.show()
```

```python
plt.figure(figsize=(10,10))
corr = df.corr(numeric_only=True)

sns.heatmap(corr, annot=False, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

```python
top_cities = df['city'].value_counts().head(10)

top_cities.plot(kind='bar')
plt.title("Top Cities")
plt.show()
```
